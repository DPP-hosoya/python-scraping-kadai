```python
import requests
from bs4 import BeautifulSoup

# 記事本文を取得する対象のURL
url = 'https://news.yahoo.co.jp/articles/1a18beba07005b731faa7ad0cbbe78b69033444e'

# GETリクエストを送信し、HTMLを取得する
response = requests.get(url)

# HTMLを解析する
soup = BeautifulSoup(response.text, 'html.parser')

# 記事本文の要素をすべて取得する
article_elements = soup.select(
    '.sc-54nboa-0.zUIkh.yjSlinkDirectlink.highLightSearchTarget'
)

# 1つ目の本文を取得
article_text1 = article_elements[0].text

# 2つ目の本文を取得
article_text2 = article_elements[1].text

# 本文を表示する
print(article_text1)
print()
print(article_text2)
```
